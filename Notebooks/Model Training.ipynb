```python
import pandas as pd
import numpy as np

df = pd.read_parquet(
    "C:/Users/venky/Desktop/Hochschule Hof/Hof Sem 2/Data Mining And Machine Learning/Data/processed/ESA-mission1/mission1_aligned.parquet"
)

print("SHAPE:", df.shape)
print("\nDTYPES:")
print(df.dtypes.value_counts())

print("\nMISSING VALUES:")
print(df.isna().sum().sum())

print("\nINFINITE VALUES:")
print(np.isinf(df.to_numpy()).sum())

print("\nDUPLICATE TIMESTAMPS:")
print(df.index.duplicated().sum())

print("\nCONSTANT CHANNELS:")
print(df.nunique()[df.nunique() <= 1])

print("\nSTATISTICAL SUMMARY:")
print(df.describe().T)

print("\nTIMESTAMP CHECK:")
print("Monotonic:", df.index.is_monotonic_increasing)
print("Frequency:", df.index.freq)
```
